```python
import numpy as np
import pandas as pd
import os
import sys
import json

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)


print("Python version:", sys.version)
print("Pandas version:", pd.__version__)
print("Numpy version:", np.__version__)
```

```python
file_path = "../data/raw/online_retail_II.xlsx"

excel_file = pd.ExcelFile(file_path)

print("Sheets Found:")
print("Sheet names:", excel_file.sheet_names)
```

```python
df_2009_2010 = pd.read_excel(
  file_path,
  sheet_name = "Year 2009-2010"
)

df_2010_2011 = pd.read_excel(
  file_path,
  sheet_name = "Year 2010-2011"
)

print("2009-2010:", df_2009_2010.shape)
print("2010-2011:", df_2010_2011.shape)
```

```python
retail = pd.concat(
    [df_2009_2010, df_2010_2011],
    ignore_index=True
)

print("Combined dataset shape:", retail.shape)
```

```python
print(retail.columns.tolist())
```

```python
retail_raw = retail.copy()

print("Raw backup created.")
print("Rows:", len(retail_raw))
```

```python
retail.head()
```

```python
retail.info()
```

```python
retail.describe(include="all").T
```

```python
missing = pd.DataFrame({
    "Missing_Count": retail.isna().sum(),
    "Missing_Percentage": retail.isna().mean() * 100
})

missing = missing.sort_values(
    "Missing_Count",
    ascending=False
)

missing
```

```python
duplicate_count = retail.duplicated().sum()

print("Duplicate rows:", duplicate_count)
print(
    "Duplicate percentage:",
    round(duplicate_count / len(retail) * 100, 2),
    "%"
)
```

```python
retail[retail.duplicated(keep=False)].head(20)
```

```python
print(retail.dtypes)
```

```python
## Convert InvoiceDate to datetime

retail["InvoiceDate"] = pd.to_datetime(
    retail["InvoiceDate"],
    errors="coerce"
)

print(retail["InvoiceDate"].dtype)
```

```python
invalid_dates = retail["InvoiceDate"].isna().sum()

print("Invalid/missing InvoiceDate:", invalid_dates)
```

```python
text_columns = [
    "Description",
    "Country",
    "StockCode"
]

for col in text_columns:
    retail[col] = retail[col].astype("string").str.strip()
```

```python
retail[["Description", "Country", "StockCode"]].head()
```

```python
retail["Invoice"] = retail["Invoice"].astype("string").str.strip()

print(retail["Invoice"].head())
```

```python
retail["IsCancelled"] = (
    retail["Invoice"]
    .str.upper()
    .str.startswith("C")
)

print(
    retail["IsCancelled"].value_counts()
)
```

```python
cancelled = retail[retail["IsCancelled"]]

print("Cancelled rows:", len(cancelled))

cancelled.head()
```

```python
cancelled["Quantity"].describe()
```

```python
## Convert Quantity to numeric
retail["Quantity"] = pd.to_numeric(
    retail["Quantity"],
    errors="coerce"
)

print(retail["Quantity"].dtype)
```

```python
## Convert UnitPrice to numeric

retail["Price"] = pd.to_numeric(
    retail["Price"],
    errors="coerce"
)
print(retail["Price"].dtype)
```

```python
## Convert CustomerID to numeric

retail["Customer ID"] = pd.to_numeric(
    retail["Customer ID"],
    errors="coerce"
)

print(retail["Customer ID"].dtype)
```

```python
print(retail["Quantity"].describe())
```

```python
negative_quantity = retail[retail["Quantity"] < 0]

print(
    "Rows with negative quantity:",
    len(negative_quantity)
)
```

```python
zero_quantity = retail[retail["Quantity"] == 0]

print(
    "Rows with zero quantity:",
    len(zero_quantity)
)
```

```python
positive_quantity = retail[retail["Quantity"] > 0]

print(
    "Rows with positive quantity:",
    len(positive_quantity)
)
```

```python
print(retail["Price"].describe())
```

```python
zero_price = retail[retail["Price"] == 0]

print("Rows with zero Price:", len(zero_price))
```

```python
negative_price = retail[retail["Price"] < 0]

print("Rows with negative Price:", len(negative_price))
```

```python
missing_customer = retail["Customer ID"].isna().sum()

print("Missing Customer ID:", missing_customer)
print(
    "Percentage:",
    round(missing_customer / len(retail) * 100, 2),
    "%"
)
```

```python
clean = retail.copy()

print("Before basic cleaning:", clean.shape)
```

```python
clean = clean.dropna(
    subset=["InvoiceDate"]
)

print(
    "After removing invalid dates:",
    clean.shape
)
```

```python
clean = clean.dropna(
    subset=["Quantity", "Price"]
)

print(
    "After removing invalid numeric values:",
    clean.shape
)
```

```python
before_duplicates = len(clean)

clean = clean.drop_duplicates()

after_duplicates = len(clean)

print("Rows before duplicates removal:", before_duplicates)
print("Rows after duplicates removal:", after_duplicates)
print(
    "Duplicates removed:",
    before_duplicates - after_duplicates
)
```

```python
print(clean.shape)
clean.head()
```

```python
clean.info()
```

```python
sales = clean[
    (~clean["IsCancelled"]) &
    (clean["Quantity"] > 0) &
    (clean["Price"] > 0) &
    (clean["Customer ID"].notna())
].copy()
```

```python
print("Sales dataset shape:", sales.shape)

sales.head()
```

```python
print(
    "Cancelled transactions remaining:",
    sales["IsCancelled"].sum()
)
```

```python
print(
    "Minimum quantity:",
    sales["Quantity"].min()
)
```

```python
print(
    "Minimum UnitPrice:",
    sales["Price"].min()
)
```

```python
print(
    "Missing Customer ID:",
    sales["Customer ID"].isna().sum()
)
```

```python
print(
    "Minimum date:",
    sales["InvoiceDate"].min()
)

print(
    "Maximum date:",
    sales["InvoiceDate"].max()
)
```

```python
unique_customers = sales["Customer ID"].nunique()

print(
    "Unique customers:",
    unique_customers
)
```

```python
unique_products = sales["StockCode"].nunique()

print(
    "Unique products:",
    unique_products
)
```

```python
unique_countries = sales["Country"].nunique()

print(
    "Unique countries:",
    unique_countries
)
```

```python
unique_invoices = sales["Invoice"].nunique()

print(
    "Unique invoices:",
    unique_invoices
)
```

```python
quality_report = {
    "raw_rows": int(len(retail_raw)),
    "clean_rows": int(len(clean)),
    "sales_rows": int(len(sales)),
    
    "raw_columns": int(retail_raw.shape[1]),
    "clean_columns": int(clean.shape[1]),
    
    "duplicate_rows_raw": int(retail_raw.duplicated().sum()),
    "duplicate_rows_clean": int(clean.duplicated().sum()),
    
    "missing_customer_id_raw": int(
        retail_raw["Customer ID"].isna().sum()
    ),
    
    "missing_customer_id_sales": int(
        sales["Customer ID"].isna().sum()
    ),
    
    "cancelled_rows": int(
        retail["IsCancelled"].sum()
    ),
    
    "negative_quantity_rows": int(
        (retail["Quantity"] < 0).sum()
    ),
    
    "zero_quantity_rows": int(
        (retail["Quantity"] == 0).sum()
    ),
    
    "zero_price_rows": int(
        (retail["Price"] == 0).sum()
    ),
    
    "unique_customers": int(
        sales["Customer ID"].nunique()
    ),
    
    "unique_products": int(
        sales["StockCode"].nunique()
    ),
    
    "unique_invoices": int(
        sales["Invoice"].nunique()
    ),
    
    "unique_countries": int(
        sales["Country"].nunique()
    ),
    
    "min_date": str(
        sales["InvoiceDate"].min()
    ),
    
    "max_date": str(
        sales["InvoiceDate"].max()
    )
}

quality_report
```

```python
output_dir = "../reports/outputs"

os.makedirs(output_dir, exist_ok=True)

report_path = os.path.join(
    output_dir,
    "data_quality_report.json"
)

with open(report_path, "w") as f:
    json.dump(
        quality_report,
        f,
        indent=4
    )

print(
    "Quality report saved:",
    report_path
)
```

```python
processed_dir = "../data/processed"

os.makedirs(processed_dir, exist_ok=True)

clean_path = os.path.join(
    processed_dir,
    "retail_clean.csv"
)

clean.to_csv(
    clean_path,
    index=False
)

print(
    "Clean transaction dataset saved:",
    clean_path
)
```

```python
sales_path = os.path.join(
    processed_dir,
    "retail_sales.csv"
)

sales.to_csv(
    sales_path,
    index=False
)

print(
    "Sales dataset saved:",
    sales_path
)
```
