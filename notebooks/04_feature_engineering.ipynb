```python
import numpy as np
import pandas as pd
import os

pd.set_option("display.max_columns", None)

print("Libraries loaded successfully..")
```

```python
sales = pd.read_csv(
    "../data/processed/retail_sales.csv"
)

sales["InvoiceDate"] = pd.to_datetime(
    sales["InvoiceDate"],
    errors="coerce"
)

print("Shape:", sales.shape)
sales.head()
```

```python
sales.info()
```

```python
## Create Revenue

sales["Revenue"] = (
  sales["Quantity"] * 
  sales["Price"]
)

sales[[
  "Quantity",
  "Price",
  "Revenue"
]].head()
```

```python
print("Minimum revenue:", sales["Revenue"].min())
print("Maximum revenue:", sales["Revenue"].max())
```

```python
## Create date features

# These will be used by:

# EDA
# forecasting
# Power BI
# Streamlit
# seasonality analysis
# demand analysis
```

```python
sales["Year"] = sales["InvoiceDate"].dt.year
sales["Month"] = sales["InvoiceDate"].dt.month
sales["MonthName"] = sales["InvoiceDate"].dt.month_name()

sales["Quarter"] = sales["InvoiceDate"].dt.quarter

sales["Day"] = sales["InvoiceDate"].dt.day
sales["DayName"] = sales["InvoiceDate"].dt.day_name()

sales["DayOfWeek"] = sales["InvoiceDate"].dt.dayofweek

sales["WeekOfYear"] = (
    sales["InvoiceDate"]
    .dt.isocalendar()
    .week
    .astype(int)
)

sales["Date"] = sales["InvoiceDate"].dt.date

sales.head()
```

```python
sales["Hour"] = sales["InvoiceDate"].dt.hour

sales["IsWeekend"] = (
    sales["DayOfWeek"] >= 5
).astype(int)

sales["IsWeekday"] = (
    sales["DayOfWeek"] < 5
).astype(int)
```

```python
sales[
    [
        "InvoiceDate",
        "Year",
        "Month",
        "Quarter",
        "DayName",
        "Hour",
        "IsWeekend"
    ]
].head(10)
```

```python
## Create Year-Month

sales["YearMonth"] = (
    sales["InvoiceDate"]
    .dt.to_period("M")
    .astype(str)
)

sales[[
    "InvoiceDate",
    "YearMonth"
]].head()

```

```python
## Create week start

sales["WeekStart"] = (
    sales["InvoiceDate"]
    .dt.to_period("W")
    .apply(lambda x: x.start_time)
)

sales[[
    "InvoiceDate",
    "WeekStart"
]].head()

```

```python
## create month satrt

sales["MonthStart"] = (
    sales["InvoiceDate"]
    .dt.to_period("M")
    .apply(lambda x: x.start_time)
)

sales[[
    "InvoiceDate",
    "MonthStart"
]].head()
```

```python
## Invoice-level features


invoice_features = (
    sales.groupby("Invoice")
    .agg(
        InvoiceQuantity=("Quantity", "sum"),
        InvoiceRevenue=("Revenue", "sum"),
        InvoiceItems=("StockCode", "nunique"),
        InvoiceCustomers=("Customer ID", "nunique")
    )
    .reset_index()
)

invoice_features.head()
```

```python
## Add invoice features back to transactions

sales = sales.merge(
    invoice_features,
    on="Invoice",
    how="left"
)

sales.head()

```

```python
## Customer-level features

# Now we create customer behavior features.

# These will later be used for:

# RFM
# customer segmentation
# churn prediction
# # customer dashboard
```

```python
customer_features = (
    sales.groupby("Customer ID")
    .agg(
        CustomerRevenue=("Revenue", "sum"),
        CustomerQuantity=("Quantity", "sum"),
        CustomerOrders=("Invoice", "nunique"),
        CustomerProducts=("StockCode", "nunique"),
        CustomerCountries=("Country", "nunique"),
        FirstPurchase=("InvoiceDate", "min"),
        LastPurchase=("InvoiceDate", "max")
    )
    .reset_index()
)

customer_features.head()
```

```python
##  Calculate average order value

# Average Order Value =
# Customer Revenue / Customer Orders


customer_features["AverageOrderValue"] = (
    customer_features["CustomerRevenue"] /
    customer_features["CustomerOrders"]
)

customer_features.head()
```

```python
## Calculate customer lifetime days

customer_features["CustomerLifetimeDays"] = (
    customer_features["LastPurchase"] -
    customer_features["FirstPurchase"]
).dt.days

customer_features.head()

```

```python
## Customer purchase frequency

customer_features["OrdersPerLifetimeDay"] = (
    customer_features["CustomerOrders"] /
    customer_features["CustomerLifetimeDays"].replace(0, np.nan)
)

customer_features["OrdersPerLifetimeDay"] = (
    customer_features["OrdersPerLifetimeDay"]
    .fillna(0)
)
```

```python
## Add customer features to transaction data

sales = sales.merge(
    customer_features,
    on="Customer ID",
    how="left"
)

print(sales.shape)
```

```python
# ## Product-level features

# Now we'll create product-level demand information.

# These will later support:

# product analysis
# demand forecasting
# inventory risk
# dashboard


product_features = (
    sales.groupby("StockCode")
    .agg(
        ProductQuantity=("Quantity", "sum"),
        ProductRevenue=("Revenue", "sum"),
        ProductOrders=("Invoice", "nunique"),
        ProductCustomers=("Customer ID", "nunique"),
        ProductFirstSale=("InvoiceDate", "min"),
        ProductLastSale=("InvoiceDate", "max")
    )
    .reset_index()
)

product_features.head()

```

```python
## Calculate product average price

product_avg_price = (
    sales.groupby("StockCode")["Price"]
    .mean()
    .reset_index(name="ProductAveragePrice")
)

product_features = product_features.merge(
    product_avg_price,
    on="StockCode",
    how="left"
)

product_features.head()
```

```python
## Add product features to sales

sales = sales.merge(
    product_features,
    on="StockCode",
    how="left"
)

print(sales.shape)
```

```python
## Calculate customer revenue percentage

total_revenue = sales["Revenue"].sum()

sales["CustomerRevenueShare"] = (
    sales["CustomerRevenue"] /
    total_revenue
)

print(
    sales["CustomerRevenueShare"].describe()
)
```

```python
## Calculate product revenue share

sales["ProductRevenueShare"] = (
    sales["ProductRevenue"] /
    total_revenue
)

sales[
    [
        "StockCode",
        "ProductRevenue",
        "ProductRevenueShare"
    ]
].head()

```

```python
## Create daily sales dataset

daily_sales = (
    sales.groupby("Date")
    .agg(
        DailyRevenue=("Revenue", "sum"),
        DailyQuantity=("Quantity", "sum"),
        DailyOrders=("Invoice", "nunique"),
        DailyCustomers=("Customer ID", "nunique"),
        DailyProducts=("StockCode", "nunique")
    )
    .reset_index()
)

daily_sales["Date"] = pd.to_datetime(
    daily_sales["Date"]
)

daily_sales.head()

```

```python
## Add daily time features

daily_sales["Year"] = daily_sales["Date"].dt.year
daily_sales["Month"] = daily_sales["Date"].dt.month
daily_sales["Quarter"] = daily_sales["Date"].dt.quarter
daily_sales["DayOfWeek"] = daily_sales["Date"].dt.dayofweek
daily_sales["WeekOfYear"] = (
    daily_sales["Date"]
    .dt.isocalendar()
    .week
    .astype(int)
)

daily_sales["IsWeekend"] = (
    daily_sales["DayOfWeek"] >= 5
).astype(int)

daily_sales.head()

```

```python
## 7-day rolling average
daily_sales = daily_sales.sort_values("Date")

daily_sales["RevenueRolling7"] = (
    daily_sales["DailyRevenue"]
    .rolling(7)
    .mean()
)
```

```python
## 14-day rolling average


daily_sales["RevenueRolling14"] = (
    daily_sales["DailyRevenue"]
    .rolling(14)
    .mean()
)
```

```python
## 30-day rolling average

daily_sales["RevenueRolling30"] = (
    daily_sales["DailyRevenue"]
    .rolling(30)
    .mean()
)
```

```python
### Create demand lag features

# Lag features are important for ML-based forecasting.

daily_sales["RevenueLag1"] = (
    daily_sales["DailyRevenue"].shift(1)
)

daily_sales["RevenueLag7"] = (
    daily_sales["DailyRevenue"].shift(7)
)

daily_sales["RevenueLag14"] = (
    daily_sales["DailyRevenue"].shift(14)
)

daily_sales["RevenueLag30"] = (
    daily_sales["DailyRevenue"].shift(30)
)
```

```python
daily_sales.tail()
```

```python
## Create monthly sales dataset

monthly_sales = (
    sales.groupby("MonthStart")
    .agg(
        MonthlyRevenue=("Revenue", "sum"),
        MonthlyQuantity=("Quantity", "sum"),
        MonthlyOrders=("Invoice", "nunique"),
        MonthlyCustomers=("Customer ID", "nunique"),
        MonthlyProducts=("StockCode", "nunique")
    )
    .reset_index()
)

monthly_sales.head()
```

```python
## Monthly growth

monthly_sales["RevenueGrowth"] = (
    monthly_sales["MonthlyRevenue"]
    .pct_change()
)

monthly_sales.head()
```

```python
## Product daily demand

# For inventory and demand forecasting, we also need product-level daily demand.

product_daily = (
    sales.groupby(
        ["Date", "StockCode"]
    )
    .agg(
        QuantitySold=("Quantity", "sum"),
        Revenue=("Revenue", "sum"),
        Orders=("Invoice", "nunique"),
        Customers=("Customer ID", "nunique")
    )
    .reset_index()
)

product_daily["Date"] = pd.to_datetime(
    product_daily["Date"]
)

product_daily.head()
```

```python
## Product demand rolling average

product_daily = product_daily.sort_values(
    ["StockCode", "Date"]
)

product_daily["DemandRolling7"] = (
    product_daily
    .groupby("StockCode")["QuantitySold"]
    .transform(
        lambda x: x.rolling(7, min_periods=1).mean()
    )
)

product_daily["DemandRolling30"] = (
    product_daily
    .groupby("StockCode")["QuantitySold"]
    .transform(
        lambda x: x.rolling(30, min_periods=1).mean()
    )
)

product_daily.head()
```

```python
## Product demand lag


product_daily["DemandLag1"] = (
    product_daily
    .groupby("StockCode")["QuantitySold"]
    .shift(1)
)

product_daily["DemandLag7"] = (
    product_daily
    .groupby("StockCode")["QuantitySold"]
    .shift(7)
)

product_daily.head(10)
```

```python
## moves unnecessary duplicate columns if any

print("Feature engineered sales shape:", sales.shape)
print("Daily sales shape:", daily_sales.shape)
print("Monthly sales shape:", monthly_sales.shape)
print("Product daily shape:", product_daily.shape)
```
