```python
import pandas as pd
import numpy as np

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
```

```python
file_path = "../data/raw/online_retail_II.xlsx"

excel_file = pd.ExcelFile(file_path)

print(excel_file.sheet_names)
```

```python
sheet_names = excel_file.sheet_names

df_1 = pd.read_excel(
    file_path,
    sheet_name=sheet_names[0]
)

df_2 = pd.read_excel(
    file_path,
    sheet_name=sheet_names[1]
)

print("Sheet 1 shape:", df_1.shape)
print("Sheet 2 shape:", df_2.shape)
```

```python
retail = pd.concat(
  [df_1, df_2],
  ignore_index=True
)
```

```python
## Missing Values Analysis
missing = retail.isnull().sum()
print("Missing Values:\n", missing)
```

```python
## Duplicate Rows Analysis
duplicate_rows = retail.duplicated().sum()
print("Duplicate Rows:", duplicate_rows)
```

```python
## Check cancellations

retail["Invoice"] = retail["Invoice"].astype(str)

cancelled = retail[
    retail["Invoice"].str.upper().str.startswith("C")
]

print("Cancelled transaction rows:", len(cancelled))
```

```python
print(retail.columns.tolist())
```

```python
verification_report = {
    "rows": retail.shape[0],
    "columns": retail.shape[1],
    "column_names": retail.columns.tolist(),
    "duplicate_rows": int(retail.duplicated().sum()),
    "missing_customer_id": int(retail["Customer ID"].isna().sum()),
    "unique_customers": int(retail["Customer ID"].nunique()),
    "unique_products": int(retail["StockCode"].nunique()),
    "unique_countries": int(retail["Country"].nunique())
}

verification_report
```

```python
import json
import os

# Define file path
file_path = "../reports/outputs/dataset_verification.json"

# Create directory path if it doesn't exist
os.makedirs(os.path.dirname(file_path), exist_ok=True)

# Write the JSON file
with open(file_path, "w") as f:
    json.dump(verification_report, f, indent=4)

print("Verification report saved successfully.")
```
